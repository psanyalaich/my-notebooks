```python
def summer_69(arr):
    
    total = 0
    add = True
    
    for num in arr:
        while add:
            if num!=6:
                total += num
                break
            else:
                add = False
        while not add:
            if num!=9:
                break
            else:
                add = True
                break
    return total
```

```python
def spy_game(nums):
    
    code = [0,0,7,'x']
    
    for num in nums:
        if num == code[0]:
            code.pop(0)
            
    return len(code) == 1
```

```python
def count_primes(num):
    
    if num < 2:
        return 0

    primes = [2]
    x = 3
    
    while x <= num:
        for y in range(3,x,2):
            if x%y == 0:
                x += 2
                break
        else:
            primes.append(x)
            x += 2
    print(primes)
    return len(primes)
```

```python
class Line:
    
    def __init__(self,coor1,coor2):
        self.coor1 = coor1
        self.coor2 = coor2
        
    def distance(self):
        
        x1,y1 = self.coor1
        x2,y2 = self.coor2
        
        return ((x2-x1)**2 + (y2-y1)**2)**0.5
        
        
    def slope(self):
        
        x1,y1 = self.coor1
        x2,y2 = self.coor2
        
        return (y2-y1) / (x2-x1)
```

```python
c1 = (3,2)
c2 = (8,10)
```

```python
myline.distance()
```

```python
myline.slope()
```

```python
class Cylinder:
    
    def __init__(self,height=1,radius=1):
        self.height = height
        self.radius = radius
        
    def volume(self):
        return self.height*3.14*(self.radius)**2
    
    def surface_area(self):
        
        top = 3.14*(self.radius)**2
        
        return (2*top)+(2*3.14*self.radius*self.height)
```

```python
import re
```

```python
pattern = r'\d{3}-\d{3}-\d{4}'
```

```python
test_string = 'here is a phone no 222-222-2222'
```

```python
re.findall(pattern,test_string)
```

```python
def search(file,pattern=r'\d{3}-\d{3}-\d{4}'):
    f = open(file,'r')
    text = f.read()
    
    if re.search(pattern,text):
        return re.search(pattern,text)
    else:
        return ''
```
